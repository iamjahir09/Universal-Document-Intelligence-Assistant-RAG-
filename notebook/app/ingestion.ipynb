Phase 1 - Data Ingestion

```python
from langchain_community.document_loaders import TextLoader,PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_google_genai import GoogleGenerativeAIEmbeddings,ChatGoogleGenerativeAI
from dotenv import load_dotenv
import pickle
import os
```

```python
document = TextLoader("../data/Neural_networks.txt",encoding='utf-8')
loaded_document = document.load()
```

```python
document_split = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50
)
splited_document = document_split.split_documents(loaded_document)

os.makedirs("../processed",exist_ok=True)
with open('../processed/chunks.pkl','wb') as f:
    pickle.dump(splited_document,f)
print("Chunks saved successfully!")

len(splited_document)
splited_document[0].page_content
splited_document[0].metadata
```

```python
from langchain_chroma import Chroma


embedding = GoogleGenerativeAIEmbeddings(model='gemini-embedding-001', google_api_key=os.getenv("GOOGLE_API_KEY"))
vectorstore = Chroma(
    collection_name="Neural",
    embedding_function=embedding,
    persist_directory="../chroma.db"
)

vectorstore.add_documents(
    documents=splited_document
)
```
